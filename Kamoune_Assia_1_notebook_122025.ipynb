```python
#Calculer la moyenne du prix
moyenne_prix = df_erp['price'].mean()
print("La moyenne du prix est :", moyenne_prix)
```

```python
import pandas as pd
from scipy import stats  # pour le calcul du z-score

#Calculer l'écart-type
ecart_type = df_erp['price'].std()
print("Écart-type du prix :", ecart_type)

#Calculer le Z-score
# Le Z-score = (valeur - moyenne) / écart-type
df_erp['z_score'] = stats.zscore(df_erp['price'])

# Afficher les premières lignes pour vérifier
print(df_erp[['price', 'z_score']].head())
```

```python
#Quel est le seuil prix dont le z-score est supérieur à 3?

# Moyenne et écart-type
moyenne = df_erp['price'].mean()
ecart_type = df_erp['price'].std()

seuil = moyenne + 3 * ecart_type
print("Seuil de prix correspondant à z-score > 3 :", seuil)
```

<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 4.2.2 - Identification par l'intervalle interquartile</h3>

```python
#Utilisation de la fonction "describe" de Pandas pour l'étude des mesures de dispersion

# Résumé statistique de la colonne 'price'
resume_prix = df_erp['price'].describe()
print(resume_prix)

```

```python
#Définir un seuil pour les articles "outliers" en prix

#Méthode de l’intervalle interquartile (IQR)
#IQR = Q3 − Q1 (quartiles 75% − 25%)
#On considère outlier :
#> Q3 + 1.5 × IQR
#< Q1 − 1.5 × IQR

Q1 = df_erp['price'].quantile(0.25)
Q3 = df_erp['price'].quantile(0.75)
IQR = Q3 - Q1

seuil_sup = Q3 + 1.5 * IQR
seuil_inf = Q1 - 1.5 * IQR

outliers = df_erp[(df_erp['price'] > seuil_sup) | (df_erp['price'] < seuil_inf)]
```

```python
#Définir le nombre d'articles et la proportion de l'ensemble du catalogue "outliers"
# Calculer Q1, Q3 et IQR
Q1 = df_erp['price'].quantile(0.25)
Q3 = df_erp['price'].quantile(0.75)
IQR = Q3 - Q1

# Définir les seuils
seuil_inf = Q1 - 1.5 * IQR
seuil_sup = Q3 + 1.5 * IQR

# Identifier les outliers
outliers_iqr = df_erp[(df_erp['price'] < seuil_inf) | (df_erp['price'] > seuil_sup)]

# Nombre et proportion
nombre_outliers_iqr = len(outliers_iqr)
proportion_outliers_iqr = nombre_outliers_iqr / len(df_erp)

print("Nombre d'articles outliers (IQR) :", nombre_outliers_iqr)
print("Proportion du catalogue (IQR) :", proportion_outliers_iqr)
```

```python
#Selon vous, ces outliers sont-ils justifiés ? Comment le démontrer si cela est possible ?

#L’analyse seule des prix ne suffit pas pour “justifier” les outliers.
#Il faut croiser avec les catégories, marques, fournisseurs et historique des ventes.
#Visualisation + Z-score + inspection métier = meilleure preuve de justification.
```

```python
######################################
# Calculer le nombre de mois de stock #
######################################

#Import de numpy 
import numpy as np

#Création de la colonne Rotation de stock

# Supposons que la rotation se calcule comme :
# Rotation = Quantité vendue / Stock quantity
# Colonne 'total_sales' pour les ventes, 'stock_quantity' pour le stock disponible

df_merge_sorted['rotation_stock'] = df_merge_sorted['total_sales'] / df_merge_sorted['stock_quantity']

# Remplacer les infinis (division par 0) par 0
df_merge_sorted['rotation_stock'] = df_merge_sorted['rotation_stock'].replace([np.inf, -np.inf], 0)

# Facultatif : remplacer les NaN (ex: stock manquant) par 0
df_merge_sorted['rotation_stock'] = df_merge_sorted['rotation_stock'].fillna(0)

# Afficher un aperçu
df_merge_sorted[['total_sales', 'stock_quantity', 'rotation_stock']].head()

#np.inf et -np.inf apparaissent si stock = 0.
#replace([np.inf, -np.inf], 0) permet de les convertir en 0.
#fillna(0) permet de gérer les valeurs manquantes (NaN) qui peuvent apparaître si le stock est NaN.

#Effectuer le tri dans l'ordre décroissant du nombre de mois de stock dans le dataset df_merge
#On calcule tout d'abord le mois_stock
df_merge_sorted['mois_stock'] = df_merge_sorted['stock_quantity'] / (df_merge_sorted['total_sales'] / 12)

#Cela donne une estimation du nombre de mois avant rupture si la tendance de vente reste constante.
#Remplacer les infinis ou NaN (stock ou ventes nulles) :
import numpy as np
df_merge_sorted['mois_stock'] = df_merge_sorted['mois_stock'].replace([np.inf, -np.inf], 0)
df_merge_sorted['mois_stock'] = df_merge_sorted['mois_stock'].fillna(0)

#Effectuer le tri dans l'ordre décroissant du nombre de mois de stock
df_merge_sorted = df_merge_sorted.sort_values(by='mois_stock', ascending=False)
flop20_stock = df_merge_sorted.head(20)

#Graphique en barre du flop 20 des produits qui ont le plus de mois de stock
fig = px.bar(
    flop20_stock,
    x='mois_stock',                   
    y='post_name',
    text='mois_stock',       
    title='Flop 20 des produits avec le plus de mois de stock',
    labels={'sku': 'SKU', 'mois_stock': 'Nombre de mois de stock'},
    color='mois_stock',        # couleur dynamique selon le stock
)

# Améliorations visuelles
fig.update_layout(
    xaxis_tickangle=-45,
    title_x=0.5,
    coloraxis_showscale=False
)

# Affichage des valeurs sur les barres
fig.update_traces(texttemplate='%{text}', textposition='outside')

fig.show()
```

```python
############################
# Analyse des corrélations #
############################

#Importation de Seaborn
import seaborn as sns

#Création d'une heatmap de corrélation avec les variables stock, sales et price
# Sélection des colonnes
cols = ['stock_quantity', 'total_sales', 'price']

# Calcul de la matrice de corrélation
corr_matrix = df_merge_sorted[cols].corr()



import matplotlib.pyplot as plt
import numpy as np
plt.figure(figsize=(8,6))
sns.heatmap(
    corr_matrix,
    annot=True,           # afficher les valeurs
    fmt=".2f",            # format des nombres
    cmap='coolwarm',      # palette de couleurs
    mask=mask,            # appliquer le mask
    cbar=True             # afficher la barre de couleur
)
plt.title("Heatmap de corrélation : stock_quantity, total_sales, price")
plt.show()

#annot=True → affiche les coefficients de corrélation directement sur la heatmap
#fmt=".2f" → deux décimales
#mask → permet de rendre la matrice plus lisible en n’affichant qu’une moitié
#cmap='coolwarm' → couleur chaude/froide pour mieux visualiser les corrélations positives/négatives

#On peut également créer un mask pour n'afficher qu'une demi heatmap
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

#np.triu → Upper Triangle (demi-matrice supérieure)
#mask sera utilisé pour cacher la moitié inférieure
```
